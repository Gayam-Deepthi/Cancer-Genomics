```python
import requests
import pandas as pd



BASE_URL = "https://www.cbioportal.org/api"
STUDY_ID = "luad_tcga_pan_can_atlas_2018"


def get_sample_list_id(study_id):
    """cBioPortal needs a 'sample list' id - usually <study_id>_all"""
    return f"{study_id}_all"

def get_mutation_molecular_profile_id(study_id):
    """
    Fetch the molecular profile ID for mutations for a given study.
    Looks for a profile of type 'MUTATION_EXT' or 'MUTATION'.
    """
    url = f"{BASE_URL}/studies/{study_id}/molecular-profiles"
    print(f"Fetching molecular profiles for study {study_id} from {url} ...")
    resp = requests.get(url)
    resp.raise_for_status()
    profiles = resp.json()

    print("Available molecular profile types and IDs:")
    for profile in profiles:
        print(f"  Type: {profile.get('molecularProfileType')}, ID: {profile.get('molecularProfileId')}")

    # Prioritize MUTATION_EXT, then MUTATION, then fall back to others if needed.
    for profile in profiles:
        if profile.get("molecularProfileType") == "MUTATION_EXT":
            print(f"Found mutation molecular profile ID: {profile['molecularProfileId']} (Type: MUTATION_EXT)")
            return profile["molecularProfileId"]
    for profile in profiles:
        if profile.get("molecularProfileType") == "MUTATION":
            print(f"Found mutation molecular profile ID: {profile['molecularProfileId']} (Type: MUTATION)")
            return profile["molecularProfileId"]
    for profile in profiles:
        # Fallback for any mutation related profile if specific types are not found
        if "mutation" in profile.get("molecularProfileId", "").lower():
            print(f"Found mutation molecular profile ID: {profile['molecularProfileId']} (Generic mutation type)")
            return profile["molecularProfileId"]

    raise ValueError(f"No suitable mutation molecular profile found for study {study_id}")


def fetch_mutations(study_id):
    """
    Fetch all mutations for the study's default mutation profile.
    Mutation profile id convention: <study_id>_mutations
    """
    molecular_profile_id = get_mutation_molecular_profile_id(study_id) # Use the dynamically found ID
    sample_list_id = get_sample_list_id(study_id)

    # Change to POST /mutations/fetch endpoint with JSON payload
    url = f"{BASE_URL}/mutations/fetch"
    payload = {
        "molecularProfileIds": [molecular_profile_id],
        "sampleListIds": [sample_list_id],
        "projection": "DETAILED" # Changed to DETAILED projection
    }

    print(f"Fetching mutations from {url} with payload {payload} ...")
    resp = requests.post(url, json=payload) # Use POST request with JSON payload
    resp.raise_for_status()

    # Check if the response text is empty before trying to parse as JSON
    if not resp.text:
        print("Warning: Received empty response body for mutations. Returning empty DataFrame.")
        return pd.DataFrame() # Return empty DataFrame

    data = resp.json()
    print(f"Retrieved {len(data)} mutation records.")
    return pd.json_normalize(data)



def fetch_clinical_data(study_id):
    """Fetch patient-level clinical data (age, sex, stage, survival, etc.)"""
    url = f"{BASE_URL}/studies/{study_id}/clinical-data"
    params = {
        "clinicalDataType": "PATIENT",
        "projection": "SUMMARY"
    }

    print(f"Fetching clinical data from {url} ...")
    resp = requests.get(url, params=params)
    resp.raise_for_status()
    data = resp.json()
    df = pd.json_normalize(data)
    print(f"Retrieved {len(df)} clinical data rows (long format).")
    return df



def pivot_clinical_data(clinical_long_df):
    """
    Clinical data comes back in long format:
    one row per (patientId, attributeId, value).
    Pivot it wide: one row per patient, one column per attribute.
    """
    wide = clinical_long_df.pivot_table(
        index="patientId",
        columns="clinicalAttributeId",
        values="value",
        aggfunc="first"
    ).reset_index()
    return wide



def main():
    # 1. Mutations
    mutations_df = fetch_mutations(STUDY_ID)
    mutations_df.to_csv("luad_mutations_raw.csv", index=False)
    print("Saved luad_mutations_raw.csv")
    if not mutations_df.empty: # Check if DataFrame is not empty before printing head
        print(mutations_df[["sampleId", "gene.hugoGeneSymbol", "proteinChange", "mutationType"]].head())
    else:
        print("No mutation data retrieved to display.")

    # 2. Clinical data
    clinical_long = fetch_clinical_data(STUDY_ID)
    clinical_long.to_csv("luad_clinical_long.csv", index=False)

    clinical_wide = pivot_clinical_data(clinical_long)
    clinical_wide.to_csv("luad_clinical_wide.csv", index=False)
    print("Saved luad_clinical_wide.csv")
    print(clinical_wide.head())


if __name__ == "__main__":
    main()
```

```python
import requests
import pandas as pd


BASE_URL = "https://www.cbioportal.org/api"
STUDY_ID = "luad_tcga_pan_can_atlas_2018"


def get_mutation_molecular_profile_id(study_id):
    """Fetch the molecular profile ID for mutations for a given study."""
    url = f"{BASE_URL}/studies/{study_id}/molecular-profiles"
    resp = requests.get(url)
    resp.raise_for_status()
    profiles = resp.json()

    for profile in profiles:
        pid = profile.get("molecularProfileId", "")
        if "mutations" in pid.lower():
            print(f"Using mutation profile: {pid}")
            return pid

    raise ValueError(f"No mutation molecular profile found for study {study_id}")


def get_best_sample_list_id(study_id):
    """
    List all sample lists for the study and pick the one most likely to
    contain samples with mutation data (prefer '_sequenced', fall back to
    '_all' only if nothing better exists).
    """
    url = f"{BASE_URL}/studies/{study_id}/sample-lists"
    resp = requests.get(url)
    resp.raise_for_status()
    sample_lists = resp.json()

    print("Available sample lists for this study:")
    for sl in sample_lists:
        print(f"  ID: {sl.get('sampleListId'):60s} | "
              f"Category: {sl.get('category')} | "
              f"#Samples: {len(sl.get('sampleIds', []))}")

    # Preferred order: sequenced samples first (these are the ones that
    # actually have mutation calls), then fall back to "all" if needed.
    preferred_keywords = ["sequenced", "acgh", "all"]

    for keyword in preferred_keywords:
        for sl in sample_lists:
            sid = sl.get("sampleListId", "").lower()
            if keyword in sid:
                print(f"-> Selected sample list: {sl['sampleListId']}")
                return sl["sampleListId"]

    # Last resort: just take the first one available
    if sample_lists:
        fallback = sample_lists[0]["sampleListId"]
        print(f"-> No preferred match found, falling back to: {fallback}")
        return fallback

    raise ValueError(f"No sample lists found for study {study_id}")


def fetch_mutations(study_id):
    molecular_profile_id = get_mutation_molecular_profile_id(study_id)
    sample_list_id = get_best_sample_list_id(study_id)

    url = f"{BASE_URL}/mutations/fetch"
    payload = {
        "molecularProfileIds": [molecular_profile_id],
        "sampleListIds": [sample_list_id],
        "projection": "DETAILED"
    }

    print(f"\nFetching mutations from {url} with payload {payload} ...")
    resp = requests.post(url, json=payload)
    resp.raise_for_status()

    if not resp.text or resp.text.strip() == "[]":
        print("Still got an empty/no-data response. See troubleshooting notes below.")
        return pd.DataFrame()

    data = resp.json()
    print(f"Retrieved {len(data)} mutation records.")
    return pd.json_normalize(data)


if __name__ == "__main__":
    mutations_df = fetch_mutations(STUDY_ID)
    mutations_df.to_csv("luad_mutations_raw.csv", index=False)
    print("\nSaved luad_mutations_raw.csv")
    if not mutations_df.empty:
        print(mutations_df[["sampleId", "gene.hugoGeneSymbol", "proteinChange", "mutationType"]].head())
    else:
        print("No mutation data retrieved — see printed sample list info above to debug further.")

        import requests
import json

BASE_URL = "https://www.cbioportal.org/api"
STUDY_ID = "luad_tcga_pan_can_atlas_2018"

# Step 1: get the list of sample lists (just metadata, no sample IDs yet)
url = f"{BASE_URL}/studies/{STUDY_ID}/sample-lists"
resp = requests.get(url)
resp.raise_for_status()
sample_lists = resp.json()

print("--- Raw JSON of the first sample list entry (from /sample-lists) ---")
print(json.dumps(sample_lists[0], indent=2))
print()








# Step 2: fetch ONE specific sample list by ID directly - this endpoint
# typically returns the full sampleIds array, unlike the summary list above.
target_id = "luad_tcga_pan_can_atlas_2018_sequenced"
url2 = f"{BASE_URL}/sample-lists/{target_id}"
resp2 = requests.get(url2)
resp2.raise_for_status()
detail = resp2.json()

print(f"--- Raw JSON of sample list '{target_id}' (from /sample-lists/{{id}}) ---")
print(json.dumps(detail, indent=2)[:2000])  # print first 2000 chars, it can be long
print(f"\nKeys present: {list(detail.keys())}")
if "sampleIds" in detail:
    print(f"Number of sampleIds: {len(detail['sampleIds'])}")



import requests
import pandas as pd

BASE_URL = "https://www.cbioportal.org/api"
STUDY_ID = "luad_tcga_pan_can_atlas_2018"
MOLECULAR_PROFILE_ID = f"{STUDY_ID}_mutations"
SAMPLE_LIST_ID = f"{STUDY_ID}_sequenced"


def get_sample_ids(study_id, sample_list_id):
    """Get the explicit list of sample IDs (not just the list name)."""
    url = f"{BASE_URL}/sample-lists/{sample_list_id}"
    resp = requests.get(url)
    resp.raise_for_status()
    detail = resp.json()
    sample_ids = detail["sampleIds"]
    print(f"Got {len(sample_ids)} explicit sample IDs from '{sample_list_id}'.")
    return sample_ids


def fetch_mutations_by_sample_ids(molecular_profile_id, sample_ids, study_id):
    """
    Fetch mutations using an explicit sampleMolecularIdentifiers list instead
    of sampleListIds. This is the payload format the POST /mutations/fetch
    endpoint expects when querying by specific samples rather than a named
    list, and tends to be more reliable.
    """
    url = f"{BASE_URL}/mutations/fetch"
    params = {"projection": "DETAILED"}

    payload = {
        "sampleMolecularIdentifiers": [
            {"molecularProfileId": molecular_profile_id, "sampleId": sid}
            for sid in sample_ids
        ]
    }

    print(f"Fetching mutations for {len(sample_ids)} samples...")
    resp = requests.post(url, params=params, json=payload)

    if resp.status_code != 200:
        print(f"Request failed ({resp.status_code}). Response body:\n{resp.text[:1000]}")
    resp.raise_for_status()

    if not resp.text or resp.text.strip() == "[]":
        print("Empty response again - see notes below.")
        return pd.DataFrame()

    data = resp.json()
    print(f"Retrieved {len(data)} mutation records.")
    return pd.json_normalize(data)


if __name__ == "__main__":
    sample_ids = get_sample_ids(STUDY_ID, SAMPLE_LIST_ID)
    mutations_df = fetch_mutations_by_sample_ids(MOLECULAR_PROFILE_ID, sample_ids, STUDY_ID)

    mutations_df.to_csv("luad_mutations_raw.csv", index=False)
    print("Saved luad_mutations_raw.csv")

    if not mutations_df.empty:
        cols_to_show = [c for c in ["sampleId", "gene.hugoGeneSymbol", "proteinChange", "mutationType"]
                         if c in mutations_df.columns]
        print(mutations_df[cols_to_show].head(10))
        print(f"\nShape: {mutations_df.shape}")
        print(f"\nColumns available: {mutations_df.columns.tolist()}")
    else:
        print("Still empty. Next step: try the bulk study download instead (see chat).")


# Step 2,3 of the project: Clean & explore the LUAD mutation data.
#
# Goal: find the most frequently mutated genes across the 566 samples,
# and sanity-check against known LUAD biology (TP53, KRAS, EGFR, STK11,
# KEAP1 should be near the top).

import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the raw mutation data you already fetched
mutations_df = pd.read_csv("luad_mutations_raw.csv")
print(f"Loaded {len(mutations_df)} mutation records.")
print(f"Number of unique samples: {mutations_df['sampleId'].nunique()}")
print(f"Number of unique genes mutated: {mutations_df['gene.hugoGeneSymbol'].nunique()}")

# 2. Mutation frequency per gene = number of UNIQUE SAMPLES with at least
#    one mutation in that gene (NOT total mutation count - a gene with
#    100 mutations in 1 sample is less interesting than a gene mutated
#    in 100 different samples).
gene_sample_counts = (
    mutations_df
    .groupby("gene.hugoGeneSymbol")["sampleId"]
    .nunique()
    .sort_values(ascending=False)
)

total_samples = mutations_df["sampleId"].nunique()
gene_freq_pct = (gene_sample_counts / total_samples * 100).round(1)

top20 = gene_freq_pct.head(20)
print("\nTop 20 most frequently mutated genes (% of samples affected):")
print(top20)

# 3. Sanity check against known LUAD driver genes
known_luad_genes = ["TP53", "KRAS", "EGFR", "STK11", "KEAP1", "NF1", "BRAF", "SETD2"]
print("\nSanity check - known LUAD driver genes and their rank/frequency:")
for gene in known_luad_genes:
    if gene in gene_freq_pct.index:
        rank = gene_freq_pct.index.get_loc(gene) + 1
        pct = gene_freq_pct[gene]
        print(f"  {gene:8s}: rank #{rank:3d}, mutated in {pct}% of samples")
    else:
        print(f"  {gene:8s}: not found in this mutation set")

# 4. Mutation type breakdown (missense vs nonsense vs etc.)
print("\nMutation type breakdown (overall):")
print(mutations_df["mutationType"].value_counts())

# 5. Visualize: bar chart of top 20 genes
plt.figure(figsize=(10, 8))
top20.sort_values().plot(kind="barh", color="darkred")
plt.xlabel("% of samples with mutation")
plt.title("Top 20 Most Frequently Mutated Genes in TCGA-LUAD (n=566 samples)")
plt.tight_layout()
plt.savefig("luad_top20_genes.png", dpi=150)
print("\nSaved chart: luad_top20_genes.png")

# 6. Save the frequency table too, useful for your report/writeup
gene_freq_pct.head(50).to_csv("luad_gene_mutation_frequency_top50.csv", header=["pct_samples_mutated"])
print("Saved luad_gene_mutation_frequency_top50.csv")
```

```python
"""
Step 4 of the project: Add clinical context - survival analysis.

Question: Does TP53 mutation status correlate with worse survival in
TCGA-LUAD patients?

We'll use Overall Survival (OS) data, which is the gold-standard endpoint
for this kind of analysis (more reliable than PFS/DFS, which depend on
follow-up scan timing rather than a hard endpoint like death).

Requirements:
    pip install pandas lifelines matplotlib
"""

import pandas as pd
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter
from lifelines.statistics import logrank_test

GENE_OF_INTEREST = "TP53"  # change this to test other genes later

# ---------------------------------------------------------------
# 1. Load your existing data
# ---------------------------------------------------------------
mutations_df = pd.read_csv("luad_mutations_raw.csv")
clinical_df = pd.read_csv("luad_clinical_wide.csv")

print(f"Mutation records: {len(mutations_df)}")
print(f"Clinical records (patients): {len(clinical_df)}")
print(f"Clinical columns available: {clinical_df.columns.tolist()}")

# ---------------------------------------------------------------
# 2. Build a patient-level TP53 mutation status table
# ---------------------------------------------------------------
# mutations_df has one row per mutation, keyed by patientId.
# We want: for each patient, is TP53 mutated (Yes/No)?
mutated_patients = set(
    mutations_df.loc[mutations_df["gene.hugoGeneSymbol"] == GENE_OF_INTEREST, "patientId"]
)
print(f"\nPatients with a {GENE_OF_INTEREST} mutation: {len(mutated_patients)}")

# All patients who have ANY mutation data (i.e. were in the sequenced cohort)
all_sequenced_patients = set(mutations_df["patientId"].unique())
print(f"Total sequenced patients: {len(all_sequenced_patients)}")

clinical_df["patientId"] = clinical_df["patientId"].astype(str)
clinical_df["tp53_status"] = clinical_df["patientId"].apply(
    lambda pid: "Mutant" if pid in mutated_patients
    else ("Wild-type" if pid in all_sequenced_patients else None)
)

# Drop patients with no mutation data at all (not in the sequenced cohort) -
# including them would bias the "wild-type" group with untested patients.
clinical_df = clinical_df[clinical_df["tp53_status"].notna()].copy()
print(f"\nPatients usable for survival analysis (have mutation data): {len(clinical_df)}")
print(clinical_df["tp53_status"].value_counts())

# ---------------------------------------------------------------
# 3. Prepare survival columns
# ---------------------------------------------------------------
# OS_MONTHS = follow-up time in months
# OS_STATUS = "1:DECEASED" or "0:LIVING" in TCGA convention
# We need: duration (numeric months) and event (1 = death, 0 = censored/alive)

print("\nChecking survival columns exist...")
required_cols = ["OS_MONTHS", "OS_STATUS"]
missing = [c for c in required_cols if c not in clinical_df.columns]
if missing:
    print(f"WARNING: missing columns {missing}. Available columns:")
    print(clinical_df.columns.tolist())
else:
    print("OS_MONTHS and OS_STATUS found.")

clinical_df["OS_MONTHS"] = pd.to_numeric(clinical_df["OS_MONTHS"], errors="coerce")
clinical_df["event"] = clinical_df["OS_STATUS"].apply(
    lambda x: 1 if isinstance(x, str) and x.startswith("1") else 0
)

# Drop rows with missing survival time - can't use them in KM analysis
before = len(clinical_df)
clinical_df = clinical_df.dropna(subset=["OS_MONTHS"])
print(f"\nDropped {before - len(clinical_df)} patients with missing OS_MONTHS.")
print(f"Final cohort for survival analysis: {len(clinical_df)} patients")

# ---------------------------------------------------------------
# 4. Kaplan-Meier survival curves: Mutant vs Wild-type
# ---------------------------------------------------------------
mutant_group = clinical_df[clinical_df["tp53_status"] == "Mutant"]
wt_group = clinical_df[clinical_df["tp53_status"] == "Wild-type"]

print(f"\nMutant group: {len(mutant_group)} patients, {mutant_group['event'].sum()} deaths observed")
print(f"Wild-type group: {len(wt_group)} patients, {wt_group['event'].sum()} deaths observed")

kmf_mutant = KaplanMeierFitter()
kmf_wt = KaplanMeierFitter()

plt.figure(figsize=(8, 6))

kmf_mutant.fit(mutant_group["OS_MONTHS"], mutant_group["event"], label=f"{GENE_OF_INTEREST} Mutant")
kmf_mutant.plot_survival_function()

kmf_wt.fit(wt_group["OS_MONTHS"], wt_group["event"], label=f"{GENE_OF_INTEREST} Wild-type")
kmf_wt.plot_survival_function()

plt.title(f"Overall Survival: {GENE_OF_INTEREST} Mutant vs Wild-type (TCGA-LUAD)")
plt.xlabel("Months")
plt.ylabel("Survival Probability")
plt.tight_layout()
plt.savefig(f"luad_{GENE_OF_INTEREST}_survival_curve.png", dpi=150)
print(f"\nSaved chart: luad_{GENE_OF_INTEREST}_survival_curve.png")

# ---------------------------------------------------------------
# 5. Statistical test: is the difference significant?
# ---------------------------------------------------------------
# Log-rank test: standard test for comparing two survival curves.
# Null hypothesis: no difference in survival between the two groups.
results = logrank_test(
    mutant_group["OS_MONTHS"], wt_group["OS_MONTHS"],
    event_observed_A=mutant_group["event"], event_observed_B=wt_group["event"]
)

print("\n--- Log-rank test results ---")
print(f"Test statistic: {results.test_statistic:.3f}")
print(f"p-value: {results.p_value:.4f}")

if results.p_value < 0.05:
    print(f"=> Statistically significant difference in survival (p < 0.05).")
else:
    print(f"=> No statistically significant difference detected (p >= 0.05).")

# Save the merged table too, useful for your report appendix
clinical_df[["patientId", "tp53_status", "OS_MONTHS", "event", "AGE",
             "AJCC_PATHOLOGIC_TUMOR_STAGE", "SEX"]].to_csv(
    f"luad_{GENE_OF_INTEREST}_survival_data.csv", index=False
)
print(f"\nSaved luad_{GENE_OF_INTEREST}_survival_data.csv")
```

```python
"""
Step 4 (extended): Compare survival impact across multiple driver genes.

Runs the same Mutant vs Wild-type log-rank test for several known LUAD
driver genes, and summarizes the results in one table - useful for a
report figure/table showing "which mutations matter most for survival".

Requirements:
    pip install pandas lifelines matplotlib
"""

import pandas as pd
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter
from lifelines.statistics import logrank_test

GENES_TO_TEST = ["TP53", "KRAS", "EGFR", "STK11", "KEAP1"]

# ---------------------------------------------------------------
# 1. Load data (same as before)
# ---------------------------------------------------------------
mutations_df = pd.read_csv("luad_mutations_raw.csv")
clinical_df = pd.read_csv("luad_clinical_wide.csv")
clinical_df["patientId"] = clinical_df["patientId"].astype(str)

all_sequenced_patients = set(mutations_df["patientId"].unique())

clinical_df["OS_MONTHS"] = pd.to_numeric(clinical_df["OS_MONTHS"], errors="coerce")
clinical_df["event"] = clinical_df["OS_STATUS"].apply(
    lambda x: 1 if isinstance(x, str) and x.startswith("1") else 0
)
clinical_df = clinical_df.dropna(subset=["OS_MONTHS"])

# Only keep patients who were actually sequenced (have known mutation status)
clinical_df = clinical_df[clinical_df["patientId"].isin(all_sequenced_patients)].copy()
print(f"Base cohort for all gene comparisons: {len(clinical_df)} patients\n")

# ---------------------------------------------------------------
# 2. Loop through each gene, run log-rank test, collect results
# ---------------------------------------------------------------
results_summary = []

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

for i, gene in enumerate(GENES_TO_TEST):
    mutated_patients = set(
        mutations_df.loc[mutations_df["gene.hugoGeneSymbol"] == gene, "patientId"]
    )

    df = clinical_df.copy()
    df["status"] = df["patientId"].apply(
        lambda pid: "Mutant" if pid in mutated_patients else "Wild-type"
    )

    mutant_group = df[df["status"] == "Mutant"]
    wt_group = df[df["status"] == "Wild-type"]

    n_mutant, n_wt = len(mutant_group), len(wt_group)
    deaths_mutant, deaths_wt = mutant_group["event"].sum(), wt_group["event"].sum()

    # Log-rank test
    test_result = logrank_test(
        mutant_group["OS_MONTHS"], wt_group["OS_MONTHS"],
        event_observed_A=mutant_group["event"], event_observed_B=wt_group["event"]
    )

    # Median survival for context (months at which survival probability = 0.5)
    kmf_m, kmf_w = KaplanMeierFitter(), KaplanMeierFitter()
    kmf_m.fit(mutant_group["OS_MONTHS"], mutant_group["event"], label=f"{gene} Mutant")
    kmf_w.fit(wt_group["OS_MONTHS"], wt_group["event"], label=f"{gene} Wild-type")

    median_mutant = kmf_m.median_survival_time_
    median_wt = kmf_w.median_survival_time_

    results_summary.append({
        "gene": gene,
        "n_mutant": n_mutant,
        "n_wildtype": n_wt,
        "pct_mutated": round(100 * n_mutant / (n_mutant + n_wt), 1),
        "deaths_mutant": int(deaths_mutant),
        "deaths_wildtype": int(deaths_wt),
        "median_survival_mutant_months": median_mutant,
        "median_survival_wt_months": median_wt,
        "p_value": round(test_result.p_value, 4),
        "significant": test_result.p_value < 0.05
    })

    # Plot into the grid
    ax = axes[i]
    kmf_m.plot_survival_function(ax=ax)
    kmf_w.plot_survival_function(ax=ax)
    ax.set_title(f"{gene} (p={test_result.p_value:.4f})")
    ax.set_xlabel("Months")
    ax.set_ylabel("Survival Probability")

# Remove the unused 6th subplot
for j in range(len(GENES_TO_TEST), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.savefig("luad_multi_gene_survival_grid.png", dpi=150)
print("Saved chart: luad_multi_gene_survival_grid.png\n")

# ---------------------------------------------------------------
# 3. Print and save summary table
# ---------------------------------------------------------------
summary_df = pd.DataFrame(results_summary).sort_values("p_value")
print("=== Summary: Survival impact by gene ===\n")
print(summary_df.to_string(index=False))

summary_df.to_csv("luad_multi_gene_survival_summary.csv", index=False)
print("\nSaved luad_multi_gene_survival_summary.csv")
```

```python
"""
Step 5 of the project: ML layer.

Goal: Predict survival outcome (deceased vs alive) using:
  - Clinical features: age, sex, tumor stage
  - Mutation status of curated driver genes: TP53, KRAS, EGFR, STK11, KEAP1

Why only 5 genes instead of all ~17,000?
  We have ~500 patients. Using thousands of gene features would massively
  outnumber our sample size, causing severe overfitting (the model would
  "memorize" patients rather than learn real patterns). Using a small,
  biologically justified set of known driver genes keeps the feature count
  sane and the model interpretable - this is standard practice in cancer
  genomics ML when sample size is limited.

Models used (matches your original project plan):
  - Logistic Regression: simple, interpretable baseline
  - Random Forest: captures non-linear interactions between features

Requirements:
    pip install pandas scikit-learn matplotlib seaborn
"""

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, roc_auc_score, classification_report,
    confusion_matrix, roc_curve
)
import matplotlib.pyplot as plt

GENES_TO_USE = ["TP53", "KRAS", "EGFR", "STK11", "KEAP1"]
RANDOM_STATE = 42  # fixed seed so results are reproducible when you re-run

# ---------------------------------------------------------------
# 1. Load data
# ---------------------------------------------------------------
mutations_df = pd.read_csv("luad_mutations_raw.csv")
clinical_df = pd.read_csv("luad_clinical_wide.csv")
clinical_df["patientId"] = clinical_df["patientId"].astype(str)

all_sequenced_patients = set(mutations_df["patientId"].unique())
clinical_df = clinical_df[clinical_df["patientId"].isin(all_sequenced_patients)].copy()

# ---------------------------------------------------------------
# 2. Build gene mutation features (1 = mutated, 0 = not mutated)
# ---------------------------------------------------------------
for gene in GENES_TO_USE:
    mutated_patients = set(
        mutations_df.loc[mutations_df["gene.hugoGeneSymbol"] == gene, "patientId"]
    )
    clinical_df[f"{gene}_mutated"] = clinical_df["patientId"].isin(mutated_patients).astype(int)

# ---------------------------------------------------------------
# 3. Build the target variable: deceased (1) vs alive (0)
# ---------------------------------------------------------------
clinical_df["deceased"] = clinical_df["OS_STATUS"].apply(
    lambda x: 1 if isinstance(x, str) and x.startswith("1") else 0
)

# ---------------------------------------------------------------
# 4. Prepare clinical features
# ---------------------------------------------------------------
clinical_df["AGE"] = pd.to_numeric(clinical_df["AGE"], errors="coerce")

# Simplify stage into a numeric ordinal: I, II, III, IV
# AJCC_PATHOLOGIC_TUMOR_STAGE looks like "STAGE IB", "STAGE IIIA" etc.
def simplify_stage(stage_str):
    if not isinstance(stage_str, str):
        return np.nan
    stage_str = stage_str.upper()
    if "IV" in stage_str:
        return 4
    elif "III" in stage_str:
        return 3
    elif "II" in stage_str:
        return 2
    elif "I" in stage_str:
        return 1
    return np.nan

clinical_df["stage_numeric"] = clinical_df["AJCC_PATHOLOGIC_TUMOR_STAGE"].apply(simplify_stage)

# Encode sex as 0/1
le_sex = LabelEncoder()
clinical_df["sex_encoded"] = le_sex.fit_transform(clinical_df["SEX"].astype(str))
print(f"Sex encoding: {dict(zip(le_sex.classes_, le_sex.transform(le_sex.classes_)))}")

# ---------------------------------------------------------------
# 5. Assemble final feature set and drop rows with missing values
# ---------------------------------------------------------------
feature_cols = ["AGE", "stage_numeric", "sex_encoded"] + [f"{g}_mutated" for g in GENES_TO_USE]

model_df = clinical_df[["patientId", "deceased"] + feature_cols].dropna()
print(f"\nPatients usable for modeling (after dropping missing values): {len(model_df)}")
print(f"Features used: {feature_cols}")
print(f"\nTarget distribution:\n{model_df['deceased'].value_counts()}")

X = model_df[feature_cols]
y = model_df["deceased"]

# ---------------------------------------------------------------
# 6. Train/test split
# ---------------------------------------------------------------
# stratify=y ensures both train and test sets have a similar proportion
# of deceased/alive patients - important with imbalanced classes.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=RANDOM_STATE, stratify=y
)
print(f"\nTrain set: {len(X_train)} patients | Test set: {len(X_test)} patients")

# Scale features for logistic regression (helps it converge properly and
# makes coefficients comparable across features of different scales)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ---------------------------------------------------------------
# 7. Train models
# ---------------------------------------------------------------
log_reg = LogisticRegression(random_state=RANDOM_STATE, max_iter=1000)
log_reg.fit(X_train_scaled, y_train)

rf = RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE, max_depth=5)
rf.fit(X_train, y_train)  # Random Forest doesn't need scaled features

# ---------------------------------------------------------------
# 8. Evaluate both models
# ---------------------------------------------------------------
def evaluate(model, X_te, y_te, name):
    preds = model.predict(X_te)
    probs = model.predict_proba(X_te)[:, 1]

    acc = accuracy_score(y_te, preds)
    auc = roc_auc_score(y_te, probs)

    print(f"\n=== {name} ===")
    print(f"Accuracy: {acc:.3f}")
    print(f"ROC-AUC:  {auc:.3f}")
    print("\nClassification report:")
    print(classification_report(y_te, preds, target_names=["Alive", "Deceased"]))

    return preds, probs, acc, auc

print("\n" + "=" * 50)
log_preds, log_probs, log_acc, log_auc = evaluate(log_reg, X_test_scaled, y_test, "Logistic Regression")
rf_preds, rf_probs, rf_acc, rf_auc = evaluate(rf, X_test, y_test, "Random Forest")

# ---------------------------------------------------------------
# 9. Feature importance (Random Forest) - which features mattered most?
# ---------------------------------------------------------------
importance_df = pd.DataFrame({
    "feature": feature_cols,
    "importance": rf.feature_importances_
}).sort_values("importance", ascending=False)

print("\n=== Random Forest Feature Importance ===")
print(importance_df.to_string(index=False))

# ---------------------------------------------------------------
# 10. Visualize: ROC curves + feature importance
# ---------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# ROC curves
for probs, name, auc in [(log_probs, "Logistic Regression", log_auc), (rf_probs, "Random Forest", rf_auc)]:
    fpr, tpr, _ = roc_curve(y_test, probs)
    axes[0].plot(fpr, tpr, label=f"{name} (AUC={auc:.3f})")
axes[0].plot([0, 1], [0, 1], "k--", alpha=0.4, label="Random guess")
axes[0].set_xlabel("False Positive Rate")
axes[0].set_ylabel("True Positive Rate")
axes[0].set_title("ROC Curves")
axes[0].legend()

# Feature importance bar chart
axes[1].barh(importance_df["feature"], importance_df["importance"], color="steelblue")
axes[1].set_xlabel("Importance")
axes[1].set_title("Random Forest Feature Importance")
axes[1].invert_yaxis()

plt.tight_layout()
plt.savefig("luad_ml_results.png", dpi=150)
print("\nSaved chart: luad_ml_results.png")

# Save predictions for reference
results_df = X_test.copy()
results_df["actual"] = y_test.values
results_df["rf_predicted"] = rf_preds
results_df["rf_probability_deceased"] = rf_probs
results_df.to_csv("luad_ml_test_predictions.csv", index=False)
print("Saved luad_ml_test_predictions.csv")
```
